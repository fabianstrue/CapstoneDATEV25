```python
import json
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from pathlib import Path
```

```python
# Load Our Filtered Data Set
file_path = Path("filtered_output.json")
with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
print(f"Number of documents in dataset: {len(data)}")
```

```python
# Flatten and count all rthema entries
all_rthemas = []

for doc in data.values():
    rthemas = doc.get("allgemeine-angaben", {}).get("rthema", [])
    all_rthemas.extend(rthemas)

# Count occurrences
rthema_counts = Counter(all_rthemas)
total = sum(rthema_counts.values())

# Convert to list of tuples (rthema, count, percentage)
rthema_stats = [
    (r, count, round((count / total) * 100, 2)) for r, count in rthema_counts.items()
]

# Sort by count descending
rthema_stats.sort(key=lambda x: x[1], reverse=True)

# Print results
print(f"{'RThema':<35} {'Count':>10} {'Percentage':>12}")
print("-" * 60)
for r, count, pct in rthema_stats:
    print(f"{r:<35} {count:>10} {pct:>11.2f}%")
```
