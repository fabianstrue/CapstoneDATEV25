```python
import os
import json
import random
import time
```

Check all documents for gruende

```python
# Input and output setup
input_path = "filtered_output.json"
output_dir = "gruende_check_output"
output_file = os.path.join(output_dir, "full_random_docs_without_gruende.txt")
os.makedirs(output_dir, exist_ok=True)

# Load data
with open(input_path, "r", encoding="utf-8") as f:
    documents = json.load(f)
# Identify documents where 'gruende' section is missing or empty
docs_without_gruende = {}

for doc_id, doc_data in documents.items():
    entscheidungsinhalt = doc_data.get("text", {}).get("entscheidungsinhalt", {})
    gruende_block = entscheidungsinhalt.get("gruende", {})

    # Accept anything non-empty inside 'gruende', regardless of its structure
    if not gruende_block or not any(gruende_block.values()):
        docs_without_gruende[doc_id] = entscheidungsinhalt

# Write all matching documents to a full text file
with open(output_file, "w", encoding="utf-8") as f_out:
    for doc_id, inhalt in docs_without_gruende.items():
        f_out.write(f"DOKUMENT-ID: {doc_id}\n")
        f_out.write("=" * 120 + "\n\n")

        for section, content in inhalt.items():
            f_out.write(f"--- {section.upper()} ---\n")

            if isinstance(content, dict):
                for key, value in content.items():
                    f_out.write(f"{key}: {value}\n")
            elif isinstance(content, list):
                for item in content:
                    f_out.write(str(item) + "\n")
            else:
                f_out.write(str(content) + "\n")

            f_out.write("\n")

        f_out.write("-" * 120 + "\n\n")

print(f"✅ Alle {len(docs_without_gruende)} Dokumente ohne gültige 'gruende' wurden gespeichert in:\n{output_file}")
```
