```python
from mistralai import Mistral

# Initialize client
api_key = "ujI60UR6Fe5jel48SAtfnMiN5Skxfwhq" 
model = "open-mistral-nemo"
client = Mistral(api_key=api_key)

from time import sleep
import json
```

```python
####ALT 
# 
# 
# 
# 
# 
# 
# # JSON-Datei mit Clustern laden
with open(r"competency_questions_output\clustered_cq_SE.json", "r", encoding="utf-8") as f:
    clusters = json.load(f)

# Ergebnis-Dictionary vorbereiten
results = {}

# Für jedes Cluster eine generalisierte Frage generieren
for cluster_id, questions in clusters.items():
    prompt = (
        "Fasse die folgenden Kompetenzfragen in eine möglichst allgemeine, repräsentative Frage zusammen:\n\n"
        + "\n".join(f"- {q}" for q in questions)
        + "\n\nAllgemeine Frage:"
    )
    try:
        response = client.chat.complete(
            model=model,
            messages=[{"role": "user", "content": prompt}]
        )
        results[cluster_id] = response.choices[0].message.content
    except Exception as e:
        print(f"Fehler bei Cluster {cluster_id}:", e)
        results[cluster_id] = "Error generating question."
    
    sleep(1)  # kurze Pause zwischen den Anfragen

# Ergebnisse speichern
with open("generalized_questions_SE.json", "w", encoding="utf-8") as f:
    json.dump(results, f, indent=2, ensure_ascii=False)

print("Fertig – generalisierte Fragen wurden gespeichert.")
```
